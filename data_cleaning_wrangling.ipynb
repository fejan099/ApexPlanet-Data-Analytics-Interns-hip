```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers_age = df[
    (df["Age"] < lower_limit) |
    (df["Age"] > upper_limit)
]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Age outliers:", len(outliers_age))
```

```python
Q1 = df["Quantity"].quantile(0.25)
Q3 = df["Quantity"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers_quantity = df[
    (df["Quantity"] < lower_limit) |
    (df["Quantity"] > upper_limit)
]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Quantity outliers:", len(outliers_quantity))
```

```python
Q1 = df["Unit_Price"].quantile(0.25)
Q3 = df["Unit_Price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers_price = df[
    (df["Unit_Price"] < lower_limit) |
    (df["Unit_Price"] > upper_limit)
]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Unit Price outliers:", len(outliers_price))
```

```python
Q1 = df["Total_Sales"].quantile(0.25)
Q3 = df["Total_Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers_sales = df[
    (df["Total_Sales"] < lower_limit) |
    (df["Total_Sales"] > upper_limit)
]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Total_Sales outliers:", len(outliers_sales))
```

```python
outliers_total_sales = df[
    (df["Total_Sales"] < lower_limit) |
    (df["Total_Sales"] > upper_limit)
]

outliers_total_sales[["Order_ID", "Order_Date", "Product", "Quantity", "Unit_Price", "Total_Sales"]]
```

```python
df["Calculated Sales"] = df["Quantity"] * df["Unit_Price"]

print((df["Total_Sales"] == df["Calculated Sales"]).all())
```

```python
mismatch = df[
    df["Total_Sales"] != df["Calculated Sales"]
]

print("Mismatched rows:", len(mismatch))
```

```python
Q1 = df["Total_Sales"].quantile(0.25)
Q3 = df["Total_Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_sales = Q1 - 1.5 * IQR
upper_sales = Q3 + 1.5 * IQR

print("Lower limit:", lower_sales)
print("Upper limit:", upper_sales)
```

```python
sales_outliers = df[
    (df["Total_Sales"] < lower_sales) |
    (df["Total_Sales"] > upper_sales)
]

print("Total_Sales outliers:", len(sales_outliers))
```

```python
sales_outliers[[
    "Order_ID",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]]
```

```python
sales_outliers["Calculated_Sales"] = (
    sales_outliers["Quantity"] * sales_outliers["Unit_Price"]
)

sales_outliers[[
    "Order_ID",
    "Quantity",
    "Unit_Price",
    "Total_Sales",
    "Calculated_Sales"
]]
```

```python
sales_outliers["Calculated_Sales"] = (sales_outliers["Quantity"] * sales_outliers["Unit_Price"])

sales_outliers[[
    "Order_ID",
    "Quantity",
    "Unit_Price",
    "Total_Sales",
    "Calculated_Sales"
]]
```

```python
sales_outliers["Sales_Check"] = (
    sales_outliers["Total_Sales"].round(2) ==
    sales_outliers["Calculated_Sales"].round(2)
)

print(sales_outliers["Sales_Check"].value_counts())
```

```python
df.drop(columns=["Calculated_Sales"], inplace=True, errors="ignore")

print(df.columns.tolist())
```
